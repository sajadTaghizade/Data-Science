# Git & GitHub Team Collaboration Guide

This handbook details how our team (**Deniz**, **Sadegh**, and **Sajjad**) will collaboratively manage code, keep work synchronized, and avoid overwriting each other's contributions.

---

## Part 1: Getting Set Up (First Time Only)

Each team member must clone the repository and switch to their designated personal sandbox branch. Open **PowerShell** and run the following commands:

```powershell
# 1. Clone the repository to your computer
git clone https://github.com/sajadTaghizade/Data-Science.git

# 2. Enter the project folder
cd .\Data-Science

# 3. Retrieve the branches created by the team lead
git fetch origin

# 4. Switch to your dedicated branch (Replace <your-name> with deniz, sadegh, or sajjad)
git switch <your-name>

```

> **Important:** Once you switch to your branch, you are safe! Any code you create, delete, or break here will not affect the rest of the team until you are ready to share it.

## Part 2: Your Daily Workflow

To ensure your local computer stays updated with the team's progress, follow this sequence when writing code or running data science scripts.

```text
 1. Pull Latest Team Changes (git pull)
               │
               ▼
 2. Write Code & Save Files 
               │
               ▼
 3. Stage Changes (git add)
               │
               ▼
 4. Save Version Locally (git commit)
               │
               ▼
 5. Upload to GitHub (git push)

```

### 1. Update your local files

Before you start typing any new code, check if the `main` branch has new updates from your teammates, and bring those updates into your branch:

```powershell
git switch main
git pull origin main
git switch <your-name>
git merge main

```

### 2. Work on your scripts

Modify your Jupyter Notebooks, write python files, or tweak documentation as usual.

### 3. Track and commit your progress

Once you complete a logical piece of work (e.g., finishing a data cleanup step), save a milestone version:

```powershell
# Check what files you changed
git status

# Stage all your changes for saving
git add .

# Save the milestone version with a brief message
git commit -m "Clean outliers from the raw dataset"

```

### 4. Push your updates to GitHub

Upload your progress to your remote branch on GitHub so it's backed up and visible to the team:

```powershell
git push origin <your-name>

```

## Part 3: Merging Your Code Into `main`

When your code or feature is finished and tested, do not try to merge it into the `main` branch inside your terminal. Instead, use GitHub's web interface via a **Pull Request (PR)**.

1. Open the repository on GitHub in your browser.
2. Click the **Pull Requests** tab at the top, then click the green **New Pull Request** button.
3. Set the base branch to `main` and the compare branch to `<your-name>`.
4. Click **Create Pull Request**, write a quick note about what you are adding, and submit it.
5. Have another team member quickly look over your code. If everything is clear, click **Merge Pull Request**.

## Part 4: Resolving Merge Conflicts (Don't Panic!)

A merge conflict sounds like a system crash, but it's actually just Git being polite. It happens when two teammates edit the exact same line of the exact same file. Git doesn’t want to guess who is right, so it pauses the merge and asks you to decide.

If you are trying to pull updates from `main` into your branch and PowerShell flashes an error saying `CONFLICT (content): Merge conflict in script.py`, here is how you fix it step-by-step.

---

### Step 1: Open the file in VS Code

Open the flagged file in your code editor. Git will have wrapped the conflicting area in weird symbols that look like this:

```text
<<<<<<< HEAD
df = df.dropna(thresh=2)  # Your local change on your branch
=======
df = df.dropna(axis=1)    # The change someone else already merged into main
>>>>>>> main

```

* **`<<<<<<< HEAD`** marks the beginning of *your* changes.
* **`=======`** is the dividing line between the two versions.
* **`>>>>>>> main`** marks the end of the changes coming from the main branch.

---

### Step 2: Choose the "Winner"

You need to clean up this messy block of code. VS Code usually gives you clickable shortcuts at the top (like *Accept Current Change* or *Accept Incoming Change*), but you can also just edit the text manually.

1. Talk to your teammate if you aren't sure whose code is better.
2. Edit the lines so only the correct code remains.
3. **Crucial:** Delete the conflict markers (`<<<<<<<`, `=======`, `>>>>>>>`) completely.

Your final code should look clean, like this:

```python
df = df.dropna(thresh=2)

```

Save the file.

---

### Step 3: Tell Git the conflict is resolved

Once your files are saved and free of those weird tracking symbols, jump back into **PowerShell** to finalize the merge:

```powershell
# 1. Stage the file to show Git you fixed it
git add script.py

# 2. Check your status to make sure everything is green
git status

# 3. Commit the merge (you can use a generic conflict message)
git commit -m "Resolved merge conflict in script.py"

# 4. Push your safely merged branch back to GitHub
git push origin <your-name>

```

> 💡 **Pro-Tip for the Team:** The absolute best way to avoid merge conflicts is to talk to each other about who is working on what file, and run `git pull origin main` at the start of every work session!
